**Theme:** Creating custom decorators <br>
**Objective:** to gain basic knowledge and practical skills in creating decorators.<br>
**Task (variant 3):** <br>
Create a custom decorator for function result caching: if a specific function has already been executed with certain parameters, do not execute the function again; instead, return the result from the cache. Test it on 3–4 functions with varying numbers and types of parameters.

**Progress of the work:** <br>
A custom decorator for caching function results was created. It stores the results of function calls and returns the cached result when the function is called with the same parameters.

```python
def cache(original_function):
    cache_data = {}
    def wrapper_function(*args, **kwargs):
        key = (args, tuple(kwargs.items()))
        if key not in cache_data:
            print("Function executed")
            cache_data[key] = original_function(*args, **kwargs)
        else:
            print("Result from cache")
        return cache_data[key]
    return wrapper_function
```

4 functions with different numbers and types of parameters were created to test the decorator. The functions have 0, 1, 2 and 3 parameters.

```python
@cache
def welcome():
    return "Welcome!"

@cache
def greet(name):
    return f"Hi, {name}!"

@cache
def add(a, b):
    return a + b

@cache
def display_info(title, author, year):
    return f"{title} by {author}, {year}"
```

The decorator was tested by calling each function twice with the same parameters. The first call executes the function, while the second call uses the cached result.

```python
print(welcome())
print(welcome())
print()
print(greet("John"))
print(greet("John"))
print()
print(add(5, 7))
print(add(5, 7))
print()
print(display_info("Harry Potter", "J. K. Rowling", 1997))
print(display_info("Harry Potter", "J. K. Rowling", 1997))
```
